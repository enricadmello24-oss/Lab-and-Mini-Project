## 1. The six comparison operators

Each one asks a question and answers with `True` or `False`.

`==` equal to · `!=` not equal to · `>` greater than · `<` less than · `>=` at least · `<=` at most

Run the next cell.

````python
value = 23.7
limit = 20

print(value == limit)
print(value > limit)
print(value <= limit)
````

### Your turn

````python
value=23.7
limit=23.7

print(value==limit)
print(value>limit)
print(value<=limit)
````

---
## 2. `=` stores, `==` asks

This is the single most common slip this week. `=` puts a value in. `==` checks one.

They read almost the same and do completely different things.

````python
limit = 20          # gets: "limit now holds 20"
print(limit == 20)  # asks: "does limit hold 20?"
````

### Your turn
Given `limit = 20`, write the line that **asks** whether `limit` is 20 — not the one that changes it.

````python
limit=20
print("Is the limit 20?",limit==20)
````

---
## 3. A comparison is a value too

The result has a type of its own: `bool`. You can store it, just like anything else.

````python
is_over = value > limit

print(is_over)
print(type(is_over))
````

### Your turn
Store the result of `87 <= 90` in a variable called `within_limit` and print it.

## 4. Combining conditions

`and` — both sides must be True. `or` — at least one side. `not` — flips it.

Bracket each comparison when you combine them. It costs nothing and removes all doubt.

````python
percent = 95.0

print((percent >= 90) and (percent < 100))
print(not (percent < 90))
````

### Your turn
Write one line that checks whether `percent` is between 90 and 100, inclusive, using `and`.

````python
percent=95.0
print(percent>=90 and percent<=100)
````

---
## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| `value = 20` when you meant to ask | nothing — it silently reassigns | use `==` to ask, `=` to store |
| `"20" == 20` | `False` | text and number are never equal — check the type |
| forgetting brackets in a combined condition | wrong answer, no error | bracket each comparison |
| `if 90 <= percent <= 100` before you are confident | works, but confusing this early | write `percent >= 90 and percent <= 100` until you are sure |
