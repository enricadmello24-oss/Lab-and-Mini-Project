## 1. `while`

Repeats the block while the condition is `True`. Same indentation rule as `if`.

````python
count = 1

while count <= 3:
    print(count)
    count += 1

print("done")
````

```
check: count <= 3 ?
   │
   ├── True  ──▶ print(count)  ──▶  count += 1  ──▶  back to check
   │
   └── False ──▶ done
```
Follow the arrows. The loop only ends when the path reaches **False** — which
is exactly why `count += 1` has to be inside the loop. Leave it out and every
lap takes the True branch, forever.

## 2. The condition must change, or it never stops

Every error so far has printed a message and stopped. An infinite loop does neither —
the program is just still running. If it happens to you: interrupt the kernel, do not
wait it out.

Read this — **do not run it**:

```python
count = 1
while count <= 3:
    print(count)
# count is never updated, so the condition is True forever
```

## 3. `while True:` + `break`

Loop until something inside the loop decides to stop, instead of counting in advance.
This is what you use when you do not know how many times you will repeat.

````python
while True:
    answer = input("Another? (y/n): ")
    if answer == "n":
        break
    print("okay, again")
````

## 4. `break` and `continue`

`break` exits the loop immediately. `continue` skips the rest of this pass and goes
straight back to checking the condition.

````python
n = 0
while n < 5:
    n += 1
    if n == 3:
        continue     # skip printing 3
    print(n)
````

### Your turn
Write a loop that asks for numbers until the user types `done`, but uses `continue` to skip adding anything when they enter a blank line.

````python
while True:
  n=input("Enter a number:")
  if n == "done":
    break
  if n == "":
    continue
````

## Common mistakes

| Mistake | What happens | Fix |
|---|---|---|
| forgetting to update what the condition checks | runs forever | make sure something inside the loop changes it |
| `while count = 5:` | `SyntaxError` | you meant `==`, or `while count < 5:` |
| `while True:` with no `break` anywhere | runs forever | every `while True:` needs an exit |
| condition checks the wrong variable | runs the wrong number of times, or never stops | trace it by hand with real numbers first |
