## 1. `if`

The indented block only runs when the condition is `True`. Nothing special happens
when it is `False` — Python just moves on.

````python
value = 23.7
limit = 20

if value > limit:
    print("OVER")
````

## 2. `else`

Runs when the `if` condition was `False`. Exactly one of the two branches runs — never
both, never neither.

````python
if value > limit:
    print("OVER")
else:
    print("OK")
````

```
check: value > limit ?
   │
   ├── True  ──▶ print("OVER")
   │
   └── False ──▶ print("OK")
```

## 3. `elif`

Checked top to bottom. The **first** `True` one runs, and the rest are skipped — even if
they are also `True`. Put the most specific condition first.

````python
percent = 105.0

if percent >= 100:
    status = "OVER LIMIT"
elif percent >= 90:
    status = "WARNING"
else:
    status = "OK"

print(status)
````

```
check: percent >= 100 ?
   │
   ├── True  ──▶ status = "OVER LIMIT"        (done - the rest is skipped)
   │
   └── False ──▶ check: percent >= 90 ?
                    │
                    ├── True  ──▶ status = "WARNING"   (done)
                    │
                    └── False ──▶ status = "OK"
```
Each `elif` only runs if every check above it was False. The moment one branch
fires, the rest are skipped — that is why order matters.

## 4. Indentation is syntax, not style

Python has no `{ }`. The indented lines **are** the block. Miss it, and Python does not
guess what you meant — it stops.

````python
if value > limit:
    print("checking...")   # part of the if
    print("OVER")           # part of the if
print("done")                # NOT indented - runs either way
````

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| missing `:` | `SyntaxError` | every `if` / `elif` / `else` line ends with `:` |
| next line not indented | `IndentationError` | indent consistently — 4 spaces |
| `if x = 5:` | `SyntaxError` | you meant `==` |
| broad condition checked before the specific one | wrong branch runs, no error | order `elif` chains most-specific first |
